```python
# if
if 5 > 10:
  print("Verdade")
```

```python
#if/else
num1 = 10
num2 = 5
if num1 > num2:
   print("O num1 é o maior número")
else:
  print("O num2 é o maior número")
```

```python
# if / Elif / Else
num1 = 10
num2 = 10
if num1 > num2:
  print("O num1 é o maior número")
elif num1 < num2:
  print("O num2 é o maior número")
else:
  print("Os dois são iguais")
```
