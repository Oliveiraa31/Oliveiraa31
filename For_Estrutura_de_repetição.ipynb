```python
For

Estrutura de repetição

Executar comandos, enquanto percorre os itens de uma coleção

Sintaxe

>>> for  <vaiável> in <coleção>:

>>>  << comandos>>



Formas de repetição

. Range

. Simples

. Enumerate



```

Estrutura de repetição

For

```python
# For Range
for contador in range(10):
  print(contador)
```

```python
# For Range + Break
for contador in range(1000):
  print(contador)
  if contador == 2:
    break
```

```python
lista = ["Teste1", "Teste2"]
```

```python
# For Simples
for va1 in lista:
   print(va1)
```

```python
# for com Enumerate
for i, va1 in enumerate(lista):
  print(i, va1)
```

For com operadores compostos

For - Operadores Compostos

Operadores Compostos

. Muito utilizado em laços de repetição

. Código mais reduzido

Operador

Mais igual / Símbolo += / Exemplo num += 1 / Equivalente num = num + 1

Menos igual / Simbolo -= / Exemplo num -= 2 / Equivalente num = num - 1

Vezes igual / Símbolo *= / Exemplo num *= 5 / Equivalente num = num * 5

Dividido igual / Símbolo /= / Exemplo num /= 2 / num = num / 2

Módulo igual /  Símbolo %= / Exemplo num %= 2 / num = num % 2



```python
lista2 = [2, 3, 4]
```

```python
for va1 in lista2:
  va1 +=1
  print(va1)
```

```python
for va1 in lista2:
  va1 *=2
  print(va1)
```
